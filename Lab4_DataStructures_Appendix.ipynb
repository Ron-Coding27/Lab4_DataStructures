```python
# Engineering Test Data Organizer

import random

LAST_NAME = "SAGCAL"
SEED_NUM = 46
FAVORITE_ARTIST = "Arthur Nery"

print(f"Last Name: {LAST_NAME}")
print(f"SEED_NUM: {SEED_NUM}")
print(f"Favorite Artist: {FAVORITE_ARTIST}")

categories = ["TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED"]
sources = [LAST_NAME + "_T" + str(i) for i in range(1, 6)]

values = []
for i, ch in enumerate(FAVORITE_ARTIST, start=1):
    values.append(SEED_NUM * i + ord(ch) % 10)

while len(values) < 5:
    values.append(SEED_NUM + len(values))

records = [
    (sources[0], categories[0], values[0]),
    (sources[1], categories[1], values[1]),
    (sources[2], categories[2], values[2]),
    (sources[3], categories[3], values[3]),
    (sources[4], categories[4], values[4]),
    (sources[0], categories[0], values[0])
]
organized = {}
for src, cat, val in records:
    if src not in organized:
        organized[src] = {}
    if cat not in organized[src]:
        organized[src][cat] = []
    organized[src][cat].append(val)

unique_records = set(records)
repeated = [rec for rec in records if records.count(rec) > 1]

numerical_summary = {
    "total_records": len(records),
    "unique_records": len(unique_records),
    "repeated_count": len(repeated)
}


print("\n=== Assessment Data ===")

print("\n1. Generated Test Data:")
print(records)

print("\n2. Organized Test Data:")
print(organized)

print("\n3. Repeated/Distinct Data:")
print("Repeated:", repeated)
print("Distinct:", unique_records)

print("\n4. Numerical and Categorical Summaries:")
print(numerical_summary)

print("\n5. Processed Results:")
print(organized)

print("\n6. Ordered Results:")
print("Sources:", sources)

print("\n7. Final Summary:")
print({
    "Seed Number": SEED_NUM,
    "Total Records": len(records),
    "Unique Records": len(unique_records),
    "Repeated Records": repeated
})
```

```python
# Student Performance Data Analyzer
import random

LAST_NAME = "SAGCAL"
SEED_NUM = 46
FAVORITE_ARTIST = "Arthur Nery"

print(f"Last Name: {LAST_NAME}")
print(f"SEED_NUM: {SEED_NUM}")
print(f"Favorite Artist: {FAVORITE_ARTIST}")

subjects = ["PROGRAMMING", "MATHEMATICS", "ELECTRONICS", "PHYSICS", "ENGLISH"]
students = [LAST_NAME + "_S" + str(i) for i in range (1, 6)]

records = [(students[i], subjects[i], (ord(subjects[i][0]) + SEED_NUM) % 100) for i in range(len(students))]
records.append(records[0])
records.append((students[2], subjects[2], None))

organized = {}
for student, subject, score in records:
    if student not in organized:
        organized[student] = {}
    organized[student][subject] = score

def categorize(score):
    if score is None:
        return "INCOMPLETE"
    elif score >= 90:
        return "EXCELLENT"
    elif score >= 75:
        return "GOOD"
    elif score >= 50:
        return "AVERAGE"
    else:
        return "POOR"

categories = [(student, subject, score, categorize(score)) for student, subject, score in records]
repeated = [rec for rec in records if records.count(rec) > 1]
incomplete = [rec for rec in records if rec[2] is None]

student_summary = {student: sum(score for subj, score in subj_dict.items() if score is not None)
                   for student, subj_dict in organized.items()}
subject_summary = {subject: [score for stu, subj, score in records if subj == subject and score is not None]
                   for subject in subjects}

combined_summary = {
    "Student Summaries": student_summary,
    "Subject Summaries": subject_summary
}

overall_value = sum(score for _, _, score in records if score is not None)

print("\n=== Assessment Data ===")
print("\n1. Generated Performance Records:")
print(records)

print("\n2. Generalized Student/Subject Data:")
print(organized)

print("\n3. Repeated/Incomplete Records:")
print("Repeated:", repeated)
print("Incomplete:", incomplete)

print("\n4. Performance Categories:")
print(categories)

print("\n5. Student and Subject Summaries:")
print(combined_summary)

print("\n6. Ordered Results:")
print("Students:", students)
print("Subjects:", subjects)



```

```python
# Transaction History Analyzer

import random

LAST_NAME = "SAGCAL"
SEED_NUM = 46
FAVORITE_ARTIST = "Arthur Nery"

print(f"Last Name: {LAST_NAME}")
print(f"SEED_NUM: {SEED_NUM}")
print(f"Favorite Artist: {FAVORITE_ARTIST}")

products = ["NOTEBOOK", "PEN", "USB", "HEADPHONES", "CALCULATOR"]
customers = [LAST_NAME + "_C" + str(i) for i in range (1, 6)]

prices = {
    "NOTEBOOK": 30,
    "PEN": 15,
    "USB": 180,
    "HEADPHONES": 350,
    "CALCULATOR": 560
}

quantities = []
for i, ch in enumerate(FAVORITE_ARTIST, start=1):
    quantities.append((SEED_NUM * i + ord(ch) % 5) % 10 + 1)

while len(quantities) < 5:
    quantities.append(SEED_NUM % 10 + 1)

records = [
    (customers[0], products[0], quantities[0], prices[products[0]]),
    (customers[1], products[1], quantities[1], prices[products[1]]),
    (customers[2], products[2], quantities[2], prices[products[2]]),
    (customers[3], products[3], quantities[3], prices[products[3]]),
    (customers[4], products[4], quantities[4], prices[products[4]]),
    (customers[0], products[0], quantities[0], prices[products[0]])
]

organized = {}
for cust, prod, qty, price in records:
    if cust not in organized:
        organized[cust] = {}
    if prod not in organized[cust]:
        organized[cust][prod] = []
    organized[cust][prod].append({"Quantity": qty, "Price": price, "Value": qty * price})
transaction_values = [(cust, prod, qty * price) for cust, prod, qty, price in records]
customer_summary = {cust: sum(qty * price for prod, data in organized[cust].items() for d in data for qty, price in [(d["Quantity"], d["Price"])]) for cust in organized}
product_summary = {prod: sum(qty * price for cust, prod2, qty, price in records if prod2 == prod) for prod in products}
overall_value = sum(qty * price for cust, prod, qty, price in records)

print("\n=== Assessment Data ===")

print("\n1. Generated Transaction Records:")
print(records)

print("\n2. Organized Transaction Data:")
print(organized)

print("\n3. Repeated Transactions:")
repeated_transactions = [rec for rec in records if records.count(rec) > 1]
print(repeated_transactions)

print("\n4. Transaction Values:")
print("Overall Value:", overall_value)

print("\n5. Customer Summaries:")
print(customer_summary)

print("\n6. Product Summaries:")
print(product_summary)

print("\n7. Ordered Results:")
print("Customers:", customers)
print("Products:", products)

print("\n8. Overall Transaction Summary:")
print({
    "Seed Number": SEED_NUM,
    "Total Records": len(records),
    "Overall Value": overall_value,
    "Repeated Purchases": [rec for rec in records if records.count(rec) > 1],
    "Distinct Customers": list(set(cust for cust, _, _, _ in records)),
    "Distinct Products": list(set(prod for _, prod, _, _ in records))
})

print("\n=== Final Consolidated Transaction Report ===")
print("Customers:", customers)
print("Products:", products)
print("Overall Value:", overall_value)
print("Repeated Purchases:", [rec for rec in records if records.count(rec) > 1])
print("Distinct Customers:", list(set(cust for cust, _, _, _ in records)))
print("Distinct Products:", list(set(prod for _, prod, _, _ in records)))
```
